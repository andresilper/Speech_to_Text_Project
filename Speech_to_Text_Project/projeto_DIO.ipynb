```python
import sounddevice as sd 
from scipy.io.wavfile import write
import numpy as np
import keyboard

fs = 44100  # taxa de amostragem
channels = 1  # mono
print("Press SPACEBAR to start recording and again to stop.")

# Espera você apertar espaço para iniciar
keyboard.wait("space")
print("Recording... press SPACEBAR to stop.")

# Lista para armazenar blocos de áudio
frames = []

# Função callback que coleta os dados do microfone
def callback(indata, frames_count, time, status):
    frames.append(indata.copy())

# Inicia stream
with sd.InputStream(samplerate=fs, channels=channels, callback=callback):
    keyboard.wait("space")  # espera você apertar espaço de novo
    print("Stop recording...")

# Concatena os blocos em um array
audio_data = np.concatenate(frames, axis=0) 

# Salva em arquivo WAV
write("my_audio.wav", fs, audio_data)
print("File saved as my_audio.wav")
```

```python
import whisper

language = 'pt'

model = whisper.load_model('small')
result = model.transcribe(r'C...\my_audio.wav', fp16=False, language=language, task='transcribe')
transcription = result["text"]
print(transcription)
```

```python
import os
from docx import Document

# Verifies if the file already exists 
if os.path.exists("my_book.docx"):
    document = Document("my_book.docx")  # opens the existing file
else:
    document = Document()                  # creates a new file
    document.add_heading("My book", 0)

# Adds the text to the transcription
document.add_paragraph(transcription)

# Saved automatically
document.save("my_book.docx")
```
